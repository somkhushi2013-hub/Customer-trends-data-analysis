###  **Customer Trends Data Analysis**

 ### Import Datasets

```python
# Loading the dataset using pandas

import pandas as pd

df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
# Summary statistics using .describe()
df.describe(include='all')
```

```python
# Checking if missing data or null values are present in the dataset

df.isnull().sum()
```

```python
# Imputing missing values in Review Rating column with the median rating of the product category

df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
# Renaming columns according to snake casing for better readability and documentation

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# Create a new column age_group based on quartiles
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']

df['age_group'] = pd.qcut(
    df['age'],
    q=4,
    labels=labels
)
```

```python
df[['age','age_group']].head(10)
```

```python
# create new column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
match_percentage = (
    (df['discount_applied'] == df['promo_code_used']).mean()
) * 100

print(f"Match Percentage: {match_percentage:.0f}%")
```

```python
# Dropping promo code used column

df = df.drop('promo_code_used', axis=1)
```

```python
import pandas as pd
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

# Step 1: Load customer dataset
dataset = pd.read_csv("customer_shopping_behavior.csv")

print("Dataset Shape:", dataset.shape)
print(dataset.head())

# Step 2: MySQL connection details
username = "root"
password = "YOUR_ACTUAL_PASSWORD"
host = "localhost"
port = 3306
database = "customer_behavior"

# Step 3: Connect to MySQL server
server_url = URL.create(
    drivername="mysql+pymysql",
    username=username,
    password=password,
    host=host,
    port=port
)

server_engine = create_engine(server_url)

# Step 4: Create database if it does not exist
with server_engine.connect() as connection:
    connection.exec_driver_sql(
        "CREATE DATABASE IF NOT EXISTS customer_behavior"
    )

print("Database created or already exists!")

# Step 5: Connect to customer_behavior database
engine = create_engine(
    URL.create(
        drivername="mysql+pymysql",
        username=username,
        password=password,
        host=host,
        port=port,
        database=database
    )
)

# Step 6: Upload dataset to MySQL
table_name = "customer"

dataset.to_sql(
    table_name,
    con=engine,
    if_exists="replace",
    index=False
)

print("Customer dataset uploaded successfully!")

# Step 7: Read sample records
result = pd.read_sql(
    "SELECT * FROM customer LIMIT 5;",
    con=engine
)

print(result)

# Step 8: Verify total records
count = pd.read_sql(
    "SELECT COUNT(*) AS total_records FROM customer;",
    con=engine
)

print(count)
```
